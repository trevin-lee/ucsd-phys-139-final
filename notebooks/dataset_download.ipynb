```python
from utils import download_tess_light_curves, download_asassn_catalog
import os

ROOT_PATH = "/Users/trevin/Git/ucsd-phys-139-final"

# --- TESS CONFIGURATION ---
SECTOR = 10           # which TESS sector (1–67+)
MAX_FILES = 20000     # how many light curves to fetch (set None for all)
TESS_DIR = f"{ROOT_PATH}/data/tess_sector{SECTOR}"

# --- GAIA CONFIGURATION ---
GAIA_DIR = f"{ROOT_PATH}/data/gaia"
GAIA_FILES = 2

# --- ASASSN CONFIGURATION ---
ASASSN_DIR = f"{ROOT_PATH}/data/asassn"
ASASSN_FILENAME = "asassn_variables_x.csv"
```

```python
# 1. Download TESS Data
import time

max_retries = 50

for attempt in range(1, max_retries + 1):
    try:
        print(f"=== Starting TESS Download (attempt {attempt}/{max_retries}) ===")
        download_tess_light_curves(SECTOR, MAX_FILES, TESS_DIR)
        print("=== TESS Download completed successfully ===")
        break
    except Exception as e:
        print(f"Download failed on attempt {attempt} with error: {e}")
        if attempt == max_retries:
            print("Max retries reached. Raising error.")
            raise
        else:
            wait_seconds = 10
            print(f"Retrying in {wait_seconds} seconds...")
            time.sleep(wait_seconds)
```

```python
# 3. Download ASASSN Data
print("\n=== Starting ASASSN Download ===")
download_asassn_catalog(ASASSN_DIR, ASASSN_FILENAME)
```
